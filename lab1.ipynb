```python
# ================================================================
# STEP 4: Data Manipulation Test (Cybersecurity Context)
# ================================================================

print("\n" + "=" * 50)
print("STEP 4: Data Manipulation Test")
print("=" * 50)

# Create sample cybersecurity log data
log_data = {
    'timestamp': ['2024-01-15 10:30:22', '2024-01-15 10:31:15', '2024-01-15 10:32:45', '2024-01-15 10:33:12'],
    'ip_address': ['192.168.1.100', '10.0.0.25', '172.16.0.50', '192.168.1.100'],
    'event_type': ['login_success', 'login_failed', 'file_access', 'login_success'],
    'user': ['alice', 'bob', 'charlie', 'alice'],
    'risk_score': [2, 8, 3, 1]
}

# Create DataFrame
df = pd.DataFrame(log_data)
print("Sample cybersecurity log data:")
print(df)

# Basic data analysis
print(f"\nDataFrame shape: {df.shape}")
print(f"High risk events (score > 5): {len(df[df['risk_score'] > 5])}")
print(f"Unique users: {df['user'].nunique()}")
print(f"Most common event type: {df['event_type'].mode().iloc[0]}")

print("✅ Step 4 Complete: Data manipulation with pandas verified!")
```

```python
# ================================================================
# STEP 5: Simple Visualization Test
# ================================================================

print("\n" + "=" * 50)
print("STEP 5: Simple Visualization Test")
print("=" * 50)

# Create a simple bar chart of event types
event_counts = df['event_type'].value_counts()

plt.figure(figsize=(8, 5))
plt.bar(event_counts.index, event_counts.values, color=['green', 'red', 'blue'])
plt.title('Cybersecurity Events by Type')
plt.xlabel('Event Type')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()

# Show the plot
plt.show()

print("✅ Step 5 Complete: Matplotlib visualization verified!")
```

```python
# ================================================================
# STEP 6: Basic Cybersecurity Operations
# ================================================================

print("\n" + "=" * 50)
print("STEP 6: Basic Cybersecurity Operations")
print("=" * 50)

# Hash a password (demonstration purposes)
password = "student_password_123"
hashed_password = hashlib.sha256(password.encode()).hexdigest()
print(f"Original password: {password}")
print(f"SHA256 hash: {hashed_password}")

# Parse a URL (useful for web security analysis)
suspicious_url = "https://example.com/login?redirect=/admin&token=abc123"
parsed_url = urllib.parse.urlparse(suspicious_url)
print(f"\nURL parsing example:")
print(f"Original URL: {suspicious_url}")
print(f"Domain: {parsed_url.netloc}")
print(f"Path: {parsed_url.path}")
print(f"Query parameters: {urllib.parse.parse_qs(parsed_url.query)}")

# Create and parse JSON (common in API security)
security_event = {
    "timestamp": "2024-01-15T10:30:22Z",
    "source_ip": "192.168.1.100",
    "event_type": "authentication",
    "success": False,
    "user_agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"
}

json_string = json.dumps(security_event, indent=2)
print(f"\nJSON security event:")
print(json_string)

parsed_event = json.loads(json_string)
print(f"\nParsed back - Event type: {parsed_event['event_type']}")

print("✅ Step 6 Complete: Basic cybersecurity operations verified!")
```
